```python
import pandas as pd
processed_df = pd.read_csv("processed_cricket_data.csv")
print(processed_df.head())
```

```python
processed_df = df.copy()

# Remove duplicate rows
processed_df = processed_df.drop_duplicates()

# Fill missing values
processed_df["Pitch_Type"] = processed_df["Pitch_Type"].fillna("Unknown")

processed_df["Net_Practice_Hours"] = processed_df["Net_Practice_Hours"].fillna(
    processed_df["Net_Practice_Hours"].mean()
)

print("Raw Rows:", len(df))
print("Processed Rows:", len(processed_df))
print(processed_df.isnull().sum())
processed_df.to_csv(
    "../Data/processed/processed_cricket_data.csv",
    index=False
)
```

```python
processed_df["Strike_Rate"] = (
    processed_df["Runs_Scored"] / processed_df["Balls_Faced"]
) * 100
```

```python
def performance_label(runs):
    if runs < 10:
        return "At Risk"
    elif runs < 30:
        return "Average"
    else:
        return "Good"

processed_df["Performance_Label"] = processed_df["Runs_Scored"].apply(performance_label)
```

```python
def experience_level(matches):
    if matches <= 20:
        return "Rookie"
    elif matches <= 100:
        return "Regular"
    else:
        return "Veteran"

processed_df["Experience_Level"] = processed_df["Previous_Matches"].apply(experience_level)
```

```python
import pandas as pd
processed_df = pd.read_csv("processed_cricket_data.csv")
print("Balls Faced vs Runs Scored:")
print(processed_df["Balls_Faced"].corr(processed_df["Runs_Scored"]))

print("Net Practice Hours vs Runs Scored:")
print(processed_df["Net_Practice_Hours"].corr(processed_df["Runs_Scored"]))

import matplotlib.pyplot as plt
import seaborn as sns

# 1. Balls Faced vs Runs Scored
plt.figure(figsize=(7, 5))
sns.regplot(
    data=processed_df,
    x="Balls_Faced",
    y="Runs_Scored",
    scatter_kws={"alpha": 0.6},
    line_kws={"color": "red"}
)
corr1 = processed_df["Balls_Faced"].corr(processed_df["Runs_Scored"])
plt.title(f"Balls Faced vs Runs Scored (r = {corr1:.2f})")
plt.xlabel("Balls Faced")
plt.ylabel("Runs Scored")
plt.tight_layout()
plt.savefig("../images/charts/balls_faced_vs_runs_scored.png", dpi=300)
plt.show()

# 2. Net Practice Hours vs Runs Scored
plt.figure(figsize=(7, 5))
sns.regplot(
    data=processed_df,
    x="Net_Practice_Hours",
    y="Runs_Scored",
    scatter_kws={"alpha": 0.6},
    line_kws={"color": "red"}
)
corr2 = processed_df["Net_Practice_Hours"].corr(processed_df["Runs_Scored"])
plt.title(f"Net Practice Hours vs Runs Scored (r = {corr2:.2f})")
plt.xlabel("Net Practice Hours")
plt.ylabel("Runs Scored")
plt.tight_layout()
plt.savefig("../images/charts/net_practice_hours_vs_runs_scored.png", dpi=300)
plt.show()
```
